## Step 1: Create a Render account

Sign up at [dashboard.render.com/register](https://dashboard.render.com/register). You can sign up with your GitHub, GitLab or Bitbucket account, or with an email address and password.

Signing up with GitHub is the simplest choice, since Render deploys your code from GitHub and you then sign in to both with one account.

Render's documentation: [Connect your Git provider](https://render.com/docs/git-provider).

## Step 2: Connect GitHub

Signing in to Render and letting Render deploy from your repositories are separate permissions, so check this even if you signed up with GitHub.

Open your [Account Settings](https://dashboard.render.com/u/settings#account-security) and scroll to **Account Security**. If GitHub isn't listed under **Git Deployment Credentials**, choose **Add credential**, pick GitHub and authorize Render on GitHub.

GitHub asks whether Render can see all your repositories or only the ones you select. Either works for the course. You can change the choice later in the Render app's settings on GitHub.

If you skip this step, Render offers to connect GitHub when you create your first web service.

Render's documentation: [Connect GitHub](https://render.com/docs/github).

## Step 3: Get to know the free tier

Nothing to set up here, but these limits explain how the free route behaves later in the week. They come from Render's page on [free instances](https://render.com/docs/free), where you can check the current details.

- **Free web services spin down when idle.** After 15 minutes without a request, Render stops the service. The next request starts it again, which takes about a minute, and Render shows a loading page meanwhile. So expect the first request of a session to be slow.
- **750 free instance hours a month, shared across your workspace.** Our services spin down when nobody is using them, so a week of labs uses far fewer hours than that. If you ever use them all, Render suspends your free services until the next month starts.
- **Nothing is kept on the service's own disk.** Free web services have no persistent disk, so anything a service writes to its disk is lost when it restarts. That's fine for us, because UnderCover keeps its data in Supabase.
- **No shell access and a single instance.** You can't open a shell on a free service, and it can't scale beyond one instance. Its logs are in the Render dashboard.
- **Use Supabase for the database.** Render offers free Postgres databases too, but they expire 30 days after creation, so we stay with the Supabase project from Week 2.
- **Payment.** If you haven't added a payment method and a free service goes over the free bandwidth or build minutes, Render suspends it or stops its builds rather than charging you. If you add one, usage beyond the free amounts is billed. The amounts are on Render's [pricing page](https://render.com/pricing).
- **Not for production.** Render says free instances shouldn't be used for production applications. They're right for a course.
